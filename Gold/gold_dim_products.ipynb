#Transformation 

```python
query = """
SELECT
    ROW_NUMBER() OVER (ORDER BY pn.product_start_date, pn.product_id) AS serial_number, -- Surrogate key
    pn.product_key,
    pn.product_id,
    pn.product_name,
    pn.category_id,
    pc.subcategory,
    pc.maintenance_flag,
    pn.product_line,
    pn.product_start_date
FROM databricks_lakehouse.silver.crm_product pn
LEFT JOIN databricks_lakehouse.silver.erp_product_catagery pc
    ON pn.category_id = pc.category_id
--WHERE pn.product_end_date IS NULL; -- removes out all historical data
"""
df = spark.sql(query)
df.display()
```

#Writing in gold

```python
(
    df.write.mode("overwrite")
    .option("overwriteSchema", "true")
    .format("delta")
    .saveAsTable("databricks_lakehouse.gold.dim_products")
)
```
