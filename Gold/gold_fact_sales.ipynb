#Transformation logic

```python
query = """
SELECT
    pr.serial_number as product_number,
    sd.order_number,
    cu.customer_key,
    sd.order_date,
    sd.ship_date,
    sd.due_date,
    sd.sales,
    sd.quantity,
    sd.price
FROM databricks_lakehouse.silver.crm_sales sd
LEFT JOIN databricks_lakehouse.gold.dim_products pr
    ON sd.product_key = pr.product_key
LEFT JOIN databricks_lakehouse.gold.dim_customers cu
    ON sd.customer_id = cu.customer_id;
"""

df = spark.sql(query)
df.display()
```

#Writting in gold

```python
(
    df.write.mode("overwrite")
    .format("delta")
    .saveAsTable("databricks_lakehouse.gold.fact_sales")
)
```
